```python
import os
from google.colab import userdata

try:
    # Try to load from Google Colab userdata (Secrets)
    grok_api_key = userdata.get("GROK_API_KEY")
except Exception:
    # Fallback to standard environment variables
    grok_api_key = os.environ.get("GROK_API_KEY")

if not grok_api_key:
    raise ValueError("Please set the GROK_API_KEY in your Colab Secrets or environment variables.")

# Install openai package if not already installed
try:
    import openai
except ImportError:
    !pip install openai
    import openai

from openai import OpenAI

# Correct base URL for Groq API
client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=grok_api_key
)

# Define the message
message = "who is best known for his starring roles in blockbuster franchises such as *Mission: Impossible*, *Top Gun*, and *Jerry Maguire*? Answer in one word"

# Request completion from Groq using a supported conversational model
response = client.chat.completions.create(
    model="qwen/qwen3.8-27b",
    messages=[
        {"role": "user", "content": message}
    ]
)

# Print the text content of the response
print(response.choices[0].message.content)
```

```python
import os

def start_chatbot():
    print("=== Groq AI CLI Chatbot ===")
    print("Type 'exit' or 'quit' to end the conversation.\n")

    # Initialize conversation history with a system message to guide the assistant
    conversation_history = [
        {"role": "system", "content": "You are a helpful, friendly, and concise AI assistant."}
    ]

    while True:
        try:
            user_input = input("You: ").strip()
        except EOFError:
            break

        if not user_input:
            continue

        if user_input.lower() in ["exit", "quit"]:
            print("Goodbye!")
            break

        # Append user message to history
        conversation_history.append({"role": "user", "content": user_input})

        try:
            # Request completion from Groq with the full conversation history
            response = client.chat.completions.create(
                model="qwen/qwen3.8-27b",
                messages=conversation_history
            )

            # Extract and print the response
            assistant_reply = response.choices[0].message.content
            print(f"\nAI: {assistant_reply}\n")

            # Append assistant response to history to maintain context
            conversation_history.append({"role": "assistant", "content": assistant_reply})

        except Exception as e:
            print(f"\nAn error occurred: {e}\n")
            # Remove the last user message if the request failed, to keep history clean
            conversation_history.pop()

# Run the chatbot
start_chatbot()
```
